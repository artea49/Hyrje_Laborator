# **Eksperimenti 1 - Variacioni natyror i kohes se ekzekutimit**

Qellimi eshte te tregohet se edhe kur kodi dhe input-i mbahen te njejte, koha e ekzekutimit nuk eshte gjithmone identike.

Do te maten 100 ekzekutime te te njejtit algoritem dhe do te llogariten:

- T_mean
- T_std
- T_min
- T_max
- CV

```python
import time
import numpy as np
import matplotlib.pyplot as plt

M = 100
N = 200_000

times = []

for r in range(M):

    s = 0

    start = time.perf_counter()

    for i in range(N):
        s += i

    end = time.perf_counter()

    times.append(end - start)

times = np.array(times)

print("T_mean =", times.mean())
print("T_std =", times.std(ddof=1))
print("T_min =", times.min())
print("T_max =", times.max())
print("CV =", times.std(ddof=1)/times.mean())

plt.hist(times,bins=20)
plt.xlabel("T [s]")
plt.ylabel("Numri i matjeve")
plt.title("Variacioni natyror i kohes")
plt.grid(True)
plt.show()
```

Rezultatet tregojne se koha e ekzekutimit nuk eshte konstante.

Edhe pse kodi dhe input-i mbahen te njejte, procesori, memoria dhe sistemi operativ krijojne variacione te vogla ne kohe.

Histogrami paraqet nje shperndarje te vlerave rreth nje mesatareje.

Devijimi standard dhe koeficienti i variacionit tregojne madhesine e ketyre ndryshimeve.

# **Eksperimenti 2 - Matje te shkurtra kundrejt matjeve te gjata**

Qellimi eshte te analizohet ndikimi i madhesise se input-it ne variacionin relativ te kohes se ekzekutimit.

```python
import time
import numpy as np

M = 100

N_values = [100,1000,10000,100000,1000000]

for N in N_values:

    times = []

    for r in range(M):

        s = 0

        start = time.perf_counter()

        for i in range(N):
            s += i

        end = time.perf_counter()

        times.append(end-start)

    times = np.array(times)

    T_mean = times.mean()
    T_std = times.std(ddof=1)
    CV = T_std/T_mean

    print("N =",N)
    print("T_mean =",T_mean)
    print("T_std =",T_std)
    print("CV =",CV)
    print()
```

Per vlera te vogla te N koha e ekzekutimit eshte shume e vogel dhe ndikohet me shume nga zhurma e sistemit.

Per vlera te medha te N ndikimi i overhead-it zvogelohet dhe matjet behen me te qendrueshme.

Prandaj CV zakonisht zvogelohet kur rritet N.

# **Eksperimenti 3 - Python kundrejt NumPy**

Qellimi eshte te krahasohet performanca e nje cikli te zakonshem Python me nje operacion te optimizuar ne NumPy.

```python
import time
import numpy as np

M = 50
N = 1000000

a = np.arange(N)

times_python = []

for r in range(M):

    s = 0

    start = time.perf_counter()

    for i in range(N):
        s += i

    end = time.perf_counter()

    times_python.append(end-start)

times_numpy = []

for r in range(M):

    start = time.perf_counter()

    s = np.sum(a)

    end = time.perf_counter()

    times_numpy.append(end-start)

times_python = np.array(times_python)
times_numpy = np.array(times_numpy)

print("Python")
print("Mean =",times_python.mean())
print("Std =",times_python.std(ddof=1))
print("CV =",times_python.std(ddof=1)/times_python.mean())

print()

print("NumPy")
print("Mean =",times_numpy.mean())
print("Std =",times_numpy.std(ddof=1))
print("CV =",times_numpy.std(ddof=1)/times_numpy.mean())
```

Rezultatet tregojne se NumPy eshte dukshem me i shpejte se nje cikel i zakonshem Python.

Kjo ndodh sepse NumPy perdor implementime te optimizuara ne kod te kompiluar, ndersa cikli Python ekzekutohet nga interpretuesi Python.

Koha mesatare e NumPy eshte me e vogel dhe ne shumicen e rasteve variacioni relativ eshte gjithashtu me i vogel.
