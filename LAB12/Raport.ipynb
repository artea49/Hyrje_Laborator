# **Eksperimenti A - Matja e fushes magnetike ambientale :**

Ne kete eksperiment telefoni u vendos larg magneteve dhe objekteve metalike. Matja u regjistrua per rreth 60 sekonda me qellim percaktimin e fushes magnetike te sfondit, e cila do te perdoret si reference ne eksperimentet e tjera.

```python
df_bg = pd.read_csv("B_background.csv")

print(df_bg.head())
print(df_bg.columns)
```

```python
df_bg = pd.read_csv("B_background.csv")

Bx = df_bg["Magnetic Field x (µT)"]
By = df_bg["Magnetic Field y (µT)"]
Bz = df_bg["Magnetic Field z (µT)"]

df_bg["B_uT"] = np.sqrt(Bx**2 + By**2 + Bz**2)

print("B mesatare =", df_bg["B_uT"].mean())
print("Devijimi standard =", df_bg["B_uT"].std(ddof=1))
```

### **Grafiku i Ekperimentit A :**

```python
plt.figure(figsize=(8,5))

plt.plot(df_bg["Time (s)"], df_bg["B_uT"])

plt.xlabel("Koha (s)")
plt.ylabel("B (uT)")
plt.title("Fusha magnetike ambientale")
plt.grid(True)

plt.show()
```

### **Analiza e Eksperimentit A**

Nga matjet e realizuara per rreth 60 sekonda u percaktua fusha magnetike ambientale e zones se matjes. Intensiteti total i fushes magnetike u llogarit duke perdorur komponentet Bx, By dhe Bz.

Vlera mesatare e fushes magnetike do te perdoret si sfond magnetik per eksperimentet pasuese. Grafiku tregon se matjet jane relativisht te qendrueshme ne kohe dhe ndryshimet e vogla lidhen me zhurmen normale te sensorit dhe ndikimet e ambientit.

```python
print("B mesatare =", df_bg["B_uT"].mean())
print("Devijimi standard =", df_bg["B_uT"].std(ddof=1))
```

### **Rezultatet e Eksperimentit A :**

Nga analiza e te dhenave u gjet:

* B mesatare = 54.23 uT
* Devijimi standard = 0.24 uT

Vlera mesatare e fushes magnetike ambientale rezultoi afersisht 54.23 uT, e cila eshte ne perputhje me rendin e madhesise se fushes magnetike te Tokes. Devijimi standard relativisht i vogel tregon se sensori ishte i qendrueshem gjate matjes dhe se kushtet e eksperimentit nuk ndryshuan ndjeshem gjate regjistrimit.

# **Eksperimenti B** - Hartezimi 1D prane magnetit

Qellimi i ketij eksperimenti ishte te studiohej ndryshimi i fushes magnetike me largesine nga magneti.

Magneti u vendos ne qender te zones se matjes dhe matjet u kryen ne pika te ndryshme pergjate nje vije te drejte. Telefoni u mbajt me te njejtin orientim ne te gjitha matjet dhe ne cdo pike u regjistruan te dhena per afersisht 5 sekonda.

```python
df1 = pd.read_csv("B_1D.csv")

print(df1.head())
print(df1.columns)
```

```python
Bx = df1["Magnetic Field x (µT)"]
By = df1["Magnetic Field y (µT)"]
Bz = df1["Magnetic Field z (µT)"]

df1["B_uT"] = np.sqrt(Bx**2 + By**2 + Bz**2)

B_background = 54.23

df1["DeltaB_uT"] = df1["B_uT"] - B_background
```

```python
plt.figure(figsize=(8,5))

plt.plot(df1["Time (s)"], df1["B_uT"])

plt.xlabel("Koha (s)")
plt.ylabel("B (uT)")
plt.title("Fusha magnetike gjate matjeve 1D")
plt.grid(True)

plt.show()
```

```python
print("B min =", df1["B_uT"].min())
print("B max =", df1["B_uT"].max())
print("B mesatare =", df1["B_uT"].mean())
```

 **Rezultatet e Eksperimentit B**

Nga analiza e te dhenave u gjet:

* B min = 13.14 uT
* B max = 52.76 uT
* B mesatare = 37.00 uT

Rezultatet tregojne se intensiteti i fushes magnetike ndryshon ne varesi te pozicionit te telefonit ndaj magnetit. Vlerat me te larta u regjistruan ne pikat me afer magnetit, ndersa me rritjen e largesise fusha magnetike u zvogelua gradualisht.

Kjo sjellje eshte ne perputhje me teorine e fushes magnetike te nje magneti, sipas se ciles intensiteti i fushes zvogelohet me largesine nga burimi.
