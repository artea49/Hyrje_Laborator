# **Teoria :**

Dendesia eshte nje madhesi fizike qe tregon raportin midis mases dhe vellimit te nje trupi.

Formula e dendesise eshte:

ρ = m / V

Per nje kub:

V = a³

ku:

- ρ eshte dendesia
- m eshte masa
- V eshte vellimi
- a eshte brinja e kubi

Meqenese dendesia varet nga masa dhe vellimi, pasigurite e ketyre madhesive ndikojne ne rezultatin perfundimtar. Per kete arsye perdoret metoda e perhapjes se pasigurive per te vleresuar saktesine e matjes

Pasiguria relative e vellimit:

$$
\frac{\Delta V}{V} = 3 \frac{\Delta a}{a}
$$

Pasiguria relative e dendesise:

$$
\frac{\Delta \rho}{\rho} =
\frac{\Delta m}{m} + \frac{\Delta V}{V}
$$

# **Te Dhenat Eksperimentale :**

```python
import numpy as np

m = np.array([42.18, 42.21, 42.20, 42.19, 42.22])

a = np.array([2.51, 2.50, 2.52, 2.51, 2.50])

print("Matjet e mases (g):")
print(m)

print("\nMatjet e brinjes (cm):")
print(a)
```

# **Perpunimi i Te Dhenave :**

```python
N = len(m)

m_mean = np.mean(m)
a_mean = np.mean(a)

sigma_m = np.std(m, ddof=1)
sigma_a = np.std(a, ddof=1)

dm = sigma_m / np.sqrt(N)
da = sigma_a / np.sqrt(N)

print("Masa mesatare =", round(m_mean,3), "g")
print("Brinja mesatare =", round(a_mean,3), "cm")

print("Pasiguria e mases =", round(dm,4), "g")
print("Pasiguria e brinjes =", round(da,4), "cm")
```

# **Analiza :**

```python
V = a_mean**3

rho = m_mean/V

rel_V = 3*(da/a_mean)

rel_rho = (dm/m_mean) + rel_V

drho = rho*rel_rho

print("Vellimi =", round(V,3), "cm^3")
print("Dendesia =", round(rho,3), "g/cm^3")
print("Pasiguria e dendesise =", round(drho,3), "g/cm^3")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Matjet e nxjerra  (në cm)
distancat_cm = [
   27.3,17.5,24.6,12.8,13,18.5,25,12.5,18.3,18,16.5,15,
13.5,9.6,21,9,10,12.5,16.9,15.3,12,11.9,11.5,18.7,
12.5,15.5,12,12,19.4,12.9,14.5,14,16.5,13.5,11.6,12.3
]

# 2. Konvertimi në metra dhe llogaritja e kohës t = sqrt(2d/g)
d = np.array(distancat_cm) / 100
g = 9.81
t = np.sqrt((2 * d) / g)

# 3. Llogaritjet statistikore
N = len(t)
mesatarja = np.mean(t)
std_dev = np.std(t, ddof=1)

print(f"Numri i matjeve: {N}")
print(f"Mesatarja (t_bar): {mesatarja:.4f} s")
print(f"Devijimi Standard (sigma): {std_dev:.4f} s")
print(f"Rezultati Final: t = ({mesatarja:.3f} ± {std_dev:.3f}) s")

# 4. Ndërtimi i grafikut (Histogrami + Kurba Gaussiane)
plt.figure(figsize=(10, 6))

# Histogrami i normalizuar 
count, bins, ignored = plt.hist(t, bins=10, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Të dhënat eksperimentale')

# Kurba teorike Gaussiane
x = np.linspace(min(t) - 0.05, max(t) + 0.05, 100)
p = norm.pdf(x, mesatarja, std_dev)
plt.plot(x, p, 'r', linewidth=2, label='Shpërndarja Normale (Gaussian)')

plt.title(' Shpërndarja e Kohës së Reagimit (Hedhja e vizores)')
plt.xlabel('Koha (s)')
plt.ylabel('Densiteti i Probabilitetit')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
