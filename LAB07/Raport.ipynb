# **Qellimi i Laboratorit:**

**Qellimi** i ketij laboratori eshte percaktimi i dendesise se nje objekti me forme kubike duke matur masen dhe dimensionet e tij.  

Ne kete eksperiment do te:

- Matim masen e objektit

- Matim brinjen e kubit

- Llogarisim vellimin

- Percaktojme dendesine

- Analizojme pasigurine e matjeve

---

# **Hyrje Teorike:**

Dendesia percaktohet nga formula:

$$
\rho = \frac{m}{V}
$$

ku:

- $\rho$ — dendesia

- $m$ — masa

- $V$ — vellimi

**Njesia e dendesise ne laborator:**

$$
g/cm^3
$$

**Per kubin, vellimi llogaritet me formulen:**

$$
V = a^3
$$

ku:

- $a$ eshte brinja e kubit

**Pasiguria relative e vellimit:**

$$
\frac{\Delta V}{V} \approx 3\frac{\Delta a}{a}
$$

**Pasiguria relative e dendesise:**

$$
\frac{\Delta \rho}{\rho}
\approx
\frac{\Delta m}{m}
+
\frac{\Delta V}{V}
$$

# **Pjesa B — Matja e Brinjes se Kubit:**

**Tabela e matjeve:**

| Matja | a (cm) |
|---|---|
| 1 | 2.51 |
| 2 | 2.50 |
| 3 | 2.52 |
| 4 | 2.51 |
| 5 | 2.50 |

**Mesatarja e brinjes:**

$$
\bar{a} =
\frac{2.51+2.50+2.52+2.51+2.50}{5}
$$

\bar{a} = 2.508 \ cm
$$

**Pasiguria e brinjes**

---

# **Pjesa C — Llogaritja e Vellimit:**

**Formula e vellimit:**

$$
V = a^3
$$

**Zevendesojme vleren mesatare:**
$$
V = (2.508)^3
$$

$$
V = 15.77 \ cm^3
$$

---


# **Pjesa D — Llogaritja e Dendesise:**

**Formula:**

$$
\rho = \frac{m}{V}
$$

$$
\rho =
\frac{42.20}{15.77}
$$

$$
\rho = 2.68 \ g/cm^3
$$

---

# **Pjesa E — Perhapja e Pasigurise:**

**Pasiguria relative e vellimit:**

$$
\frac{\Delta V}{V} =
3\frac{\Delta a}{a}
$$

$$
\frac{\Delta V}{V} =
3 \cdot \frac{0.01}{2.508}
$$

$$
\frac{\Delta V}{V}
\approx
0.012
$$

**Pasiguria relative e dendesise:**

$$
\frac{\Delta \rho}{\rho} =
\frac{\Delta m}{m}
+
\frac{\Delta V}{V}
$$

$$
\frac{\Delta \rho}{\rho} =
\frac{0.02}{42.20}
+
0.012
$$

$$
\frac{\Delta \rho}{\rho}
\approx
0.0125
$$

**Pasiguria absolute e dendesise:**

$$
\Delta \rho =
\rho \cdot \frac{\Delta \rho}{\rho}
$$

$$
\Delta \rho =
2.68 \cdot 0.0125
$$

$$
\Delta \rho \approx 0.03 \ g/cm^3
$$

---



# **Rezultati Final:**

$$
\boxed{
\rho = (2.68 \pm 0.03)\ g/cm^3
}
$$
