# **Eksperimenti 1 – Algoritmi me nje cikel**

Algoritmi perbehet nga nje cikel i vetem for qe llogarit shumen e numrave nga 0 deri ne N.

Kompleksiteti teorik:

T(N) ~ N

Prandaj pritet qe eksponenti i matur te jete afersisht:

p ≈ 1

```python
import time
import numpy as np

def experiment1(N, reps=10):
    times = []

    for _ in range(reps):
        s = 0

        start = time.perf_counter()

        for i in range(N):
            s += i

        end = time.perf_counter()

        times.append(end - start)

    return np.mean(times), np.std(times, ddof=1)

N_values1 = [10, 100, 1000, 10000, 100000, 1000000]

results1 = [experiment1(N) for N in N_values1]

for N, (mean, std) in zip(N_values1, results1):
    print(f"N={N}, T_mean={mean:.8f}s, T_std={std:.8f}s")
```

```python
import matplotlib.pyplot as plt

means1 = [r[0] for r in results1]

plt.loglog(N_values1, means1, 'o-')
plt.xlabel("N")
plt.ylabel("T_mean [s]")
plt.title("Eksperimenti 1")
plt.grid(True)
plt.show()
```

```python
x = np.log10(N_values1)
y = np.log10(means1)

p, b = np.polyfit(x, y, 1)

print("p =", p)
```

## **Eksperimenti 2 - Dy cikle te futura**

**Qellimi :** Te matet koha e ekzekutimit e nje algoritmi me dy cikle te futura dhe te studiohet varesia e saj nga madhesia e inputit N.

**Algoritmi :** Algoritmi perdor dy cikle te futura:

for i in range(N):

    for j in range(N):

        s += i * j

Kompleksiteti teorik i pritshem eshte:

T
(
N
)
∼
N
2

```python
import time
import numpy as np

def experiment2(N, reps=10):
    times = []

    for _ in range(reps):
        s = 0

        start = time.perf_counter()

        for i in range(N):
            for j in range(N):
                s += i * j

        end = time.perf_counter()

        times.append(end - start)

    return np.mean(times), np.std(times)

N_values2 = [10, 30, 100, 300, 1000]

results2 = [experiment2(N) for N in N_values2]

for N, (mean, std) in zip(N_values2, results2):
    print(f"N={N}, T_mean={mean:.6f}s, T_std={std:.6f}s")
```

```python
import matplotlib.pyplot as plt

T_mean2 = [r[0] for r in results2]

plt.figure(figsize=(6,4))
plt.loglog(N_values2, T_mean2, "o-")
plt.xlabel("N")
plt.ylabel("T_mean [s]")
plt.title("Eksperimenti 2 - Dy cikle")
plt.grid(True)
plt.show()
```

```python
x = np.log10(N_values2)
y = np.log10(T_mean2)

p, b = np.polyfit(x, y, 1)

print("p =", p)
print("b =", b)
```

## **Eksperimenti 3 - Tre cikle te futura**

**Qellimi :** Te matet koha e ekzekutimit per nje algoritem me tre cikle te futura.

**Algoritmi :**

for i in range(N):
    for j in range(N):
        for k in range(N):
            s += i * j * k

Kompleksiteti teorik:

T(N) ~ N³


```python
import time
import numpy as np

def experiment3(N, reps=10):
    times = []

    for _ in range(reps):
        s = 0

        start = time.perf_counter()

        for i in range(N):
            for j in range(N):
                for k in range(N):
                    s += i * j * k

        end = time.perf_counter()

        times.append(end - start)

    return np.mean(times), np.std(times)

N_values3 = [5, 10, 20, 30, 50]

results3 = [experiment3(N) for N in N_values3]

for N, (mean, std) in zip(N_values3, results3):
    print(f"N={N}, T_mean={mean:.6f}s, T_std={std:.6f}s")
```

```python
T_mean3 = [r[0] for r in results3]

plt.figure(figsize=(6,4))
plt.loglog(N_values3, T_mean3, "o-")
plt.xlabel("N")
plt.ylabel("T_mean [s]")
plt.title("Eksperimenti 3 - Tre cikle")
plt.grid(True)
plt.show()
```

```python
x = np.log10(N_values3)
y = np.log10(T_mean3)

p, b = np.polyfit(x, y, 1)

print("p =", p)
print("b =", b)
```

**Analiza :** Grafiku log-log jep nje vije afersisht lineare.

Nga fitimi linear u gjet:

p ≈ 3

Rezultati perputhet me kompleksitetin teorik O(N³).

## Eksperimenti 4 - Renditja me NumPy

**Qellimi :** Te matet koha e ekzekutimit e algoritmit te renditjes se nje vargu me NumPy dhe te krahasohet me kompleksitetin teorik.

**Algoritmi :** Per cdo vlere te N krijohet nje varg me numra te rastesishem dhe renditet me funksionin:

np.sort()

Kompleksiteti teorik i pritshem eshte:

T(N) ~ N log N



```python
import time
import numpy as np

def experiment4(N, reps=10):
    times = []

    for _ in range(reps):

        a = np.random.rand(N)

        start = time.perf_counter()

        b = np.sort(a)

        end = time.perf_counter()

        times.append(end - start)

    return np.mean(times), np.std(times)

N_values4 = [100, 1000, 10000, 100000, 1000000]

results4 = [experiment4(N) for N in N_values4]

for N, (mean, std) in zip(N_values4, results4):
    print(f"N={N}, T_mean={mean:.6f}s, T_std={std:.6f}s")
```

```python
import matplotlib.pyplot as plt

T_mean4 = [r[0] for r in results4]

plt.figure(figsize=(6,4))
plt.loglog(N_values4, T_mean4, "o-")

plt.xlabel("N")
plt.ylabel("T_mean [s]")
plt.title("Eksperimenti 4 - NumPy Sort")
plt.grid(True)

plt.show()
```

```python
x = np.log10(N_values4)
y = np.log10(T_mean4)

p, b = np.polyfit(x, y, 1)

print("p =", p)
print("b =", b)
```

**Rezultatet :** U mat koha mesatare e renditjes per madhesi te ndryshme te inputit.

**Analiza :** Grafiku log-log tregon se koha rritet me shpejt se nje varesi lineare, por me ngadale se nje varesi kuadratike.

Eksponenti i marre nga fitimi linear nuk eshte saktesisht 1 ose 2, gje qe eshte ne perputhje me kompleksitetin teorik O(N log N).
