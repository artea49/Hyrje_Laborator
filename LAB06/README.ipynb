# LAB 06 – Matja e Madhesive Baze: Gjatesia dhe Analiza e Pasigurise
---

## Pershkrimi i projektit:

Ky laborator ka si qellim matjen eksperimentale te madhesive baze fizike (gjatesise) duke perdorur instrumente te ndryshme si vizore, kaliper vernier dhe sferometer.

Eksperimenti fokusohet ne mbledhjen e te dhenave te perseritura dhe analizen statistikore te tyre per te vleresuar pasigurite matese.

Per secilin objekt (bateri, suste dhe xham) u kryen disa matje te perseritura per te analizuar:
- precizionin e instrumenteve
- gabimet rastesore dhe sistematike
- ndikimin e rezolucionit ne saktesi

---

## Struktura e projektit:

- `data/` → te dhenat eksperimentale (battery.dat, spring.dat, glass.dat)
- `notebook.ipynb` → analiza dhe llogaritjet ne Python
- `README.md` → pershkrimi i laboratorit
- `Reports/` → raporti final

---

## Pajisjet e perdorura:

- Vizore (rezolucion i ulet)
- Kaliper vernier (rezolucion i larte)
- Sferometer (për matje shume te vogla)
- Objekte: bateri, suste, xham

---

## Konceptet kryesore:

- Matja fizike dhe pasiguria
- Precizioni dhe rezolucioni
- Gabimet sistematike dhe rastesore
- Analiza statistikore e te dhenave

Matja shprehet si:

$$
x = x_{\text{mes}} \pm \Delta x
$$
